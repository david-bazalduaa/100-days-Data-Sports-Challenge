## Technical Retrospective (System Design Q&A)

### Q1: Performance & Storage (Caching and Parquet)
**Why is local caching (`_check_local_cache`) and Parquet format critical before entering Phase 3 (Modeling)?**
*Answer:* In Machine Learning, feature engineering and hyperparameter tuning require hundreds of iterative runs. Relying on network calls to public APIs (StatsBomb/nflverse) for every run introduces massive latency, risks hitting rate limits, and creates a dependency on external server uptime. Saving data locally in `.parquet` format (using `pyarrow`) compresses data efficiently and preserves strict data types (like nested arrays for tracking coordinates), allowing for lightning-fast, reproducible, and offline experimentation.

### Q2: Data Resilience (Quality Validation)
**How does our current architecture protect our future ML models from failing silently if upstream data changes unexpectedly (e.g., column name changes or missing `x,y` coordinates)?**
*Answer:* By implementing strict Data Quality Checks and defining data schemas, we established a "fail-fast" pipeline. If upstream providers alter column names or inject nulls into critical features, our ingestion layer will catch the anomaly and raise an explicit error before the data reaches the model. This prevents "garbage-in, garbage-out" (GIGO) scenarios, ensuring our models don't silently train on corrupted data, which would severely degrade predictive performance in production.

### Q3: Portability (Dynamic Pathing)
**Beyond switching between personal laptops, why is strict dynamic pathing (`os.path.abspath`) mandatory for production environments?**
*Answer:* Hardcoded absolute paths tie the codebase to a specific machine and operating system. In production (Phase 4), our application will be deployed in containerized environments like Docker or cloud-based Linux servers (AWS/GCP). Using dynamic, relative paths ensures the application is completely environment-agnostic, enabling seamless CI/CD workflows and preventing fatal `PermissionError` or `FileNotFoundError` exceptions when the code runs in the cloud.
